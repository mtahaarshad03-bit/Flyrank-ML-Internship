## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

### Unit of Analysis: One row = One page / search query

**Definition:** Each row in this dataset represents an individual search query or content URL, along with its performance metrics (like impression count, ranking position, and click-through rate).

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
import glob

# Search for the starter CSV dataset inside repo folders
file_paths = glob.glob('**/*.csv', recursive=True)

if file_paths:
    # Load dataset
    df = pd.read_csv(file_paths[0])
    print(f"Loaded File: {file_paths[0]}")
    print("Unit of Analysis: One row = One search query / page URL\n")

    # 1. Show Dataframe Preview
    display(df.head())

    # 2. Sketch Target Column (1 if position <= 10 else 0)
    if 'position' in df.columns:
        df['is_high_performing'] = (df['position'] <= 10).astype(int)
        print("\nTarget Column Preview:")
        display(df[['position', 'is_high_performing']].head())
    else:
        # Fallback if position column isn't directly named 'position'
        df['is_high_performing'] = 1
        display(df.head())
else:
    print("No CSV found. Please verify dataset path.")
```
